```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score
```

```python
data = {
    'Nim': ['TI001', 'TI002', 'TI003', 'TI004', 'TI005', 'TI006', 'TI007', 'TI008', 'TI009', 'TI010'],
    'Hasil_Sebenarnya': ['Lulus', 'Lulus', 'Lulus', 'Lulus', 'Lulus', 'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus'],
    'Hasil_Prediksi':   ['Lulus', 'Lulus', 'Lulus', 'Tidak Lulus', 'Tidak Lulus', 'Lulus', 'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus', 'Tidak Lulus']
}

df = pd.DataFrame(data)

print("DATA HASIL PREDIKSI KNN:")
print(df)
print("\n" + "="*60 + "\n")
```

```python
labels = ['Lulus', 'Tidak Lulus']
y_true = df['Hasil_Sebenarnya']
y_pred = df['Hasil_Prediksi']

cm = confusion_matrix(y_true, y_pred, labels=labels)
```

```python
plt.figure(figsize=(6, 5))

sns.heatmap(cm, annot=True, fmt='d', cmap='Greens', xticklabels=labels, yticklabels=labels)

plt.xlabel('Hasil Prediksi Model')
plt.ylabel('Hasil Sebenarnya (Aktual)')
plt.title('Confusion Matrix Evaluasi Model KNN')
plt.show()

print("\n" + "="*60 + "\n")
```

```python
accuracy = accuracy_score(y_true, y_pred)
precision = precision_score(y_true, y_pred, pos_label='Lulus')
recall = recall_score(y_true, y_pred, pos_label='Lulus')

print("HASIL PERHITUNGAN PERSENTASE:")
print(f"1. Accuracy (Tingkat Akurasi)  : {accuracy * 100:.0f}%")
print(f"2. Precision (Ketepatan Prediksi): {precision * 100:.0f}%")
print(f"3. Recall (Daya Ingat Model)     : {recall * 100:.0f}%")
```
